```python
rank_points_by_year = []

for f in files_1985:
    year = int(f.stem.split("_")[-1])
    df_tmp = pd.read_csv(f, usecols=["winner_rank_points", "loser_rank_points"])
    missing_w = df_tmp["winner_rank_points"].isna().sum()
    missing_l = df_tmp["loser_rank_points"].isna().sum()
    total = len(df_tmp)

    rank_points_by_year.append({
        "year": year,
        "rows": total,
        "winner_rank_points_missing": missing_w,
        "loser_rank_points_missing": missing_l,
        "winner_missing_pct": missing_w / total,
        "loser_missing_pct": missing_l / total,
    })

rank_points_year_df = pd.DataFrame(rank_points_by_year)
rank_points_year_df.head(15)
```
