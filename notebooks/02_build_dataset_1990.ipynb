```python
import pandas as pd
from pathlib import Path
```

```python
DATA_DIR = Path("../data/raw/sackmann")
INTERIM_DIR = Path("../data/interim")
INTERIM_DIR.mkdir(parents = True, exist_ok = True)
```

```python
files = sorted(DATA_DIR.glob("atp_matches_*.csv"))

files_main = []
for f in files:
    parts = f.stem.split("_")
    if len(parts) == 3 and parts[0] == "atp" and parts[1] == "matches" and parts[2].isdigit():
        files_main.append(f)

files_main = sorted(files_main)
len(files_main), files_main[0].name, files_main[-1].name
```

```python
files_1990 = [f for f in files_main if int(f.stem.split("_")[-1]) >= 1990]
len(files_1990), files_1990[0].name, files_1990[-1].name
```

```python
KEEP_COLS = [
    "tourney_id",
    "tourney_name",
    "surface",
    "tourney_level",
    "tourney_date",
    "match_num",
    "winner_id",
    "winner_name",
    "winner_age",
    "loser_id",
    "loser_name",
    "loser_age",
    "best_of",
    "winner_rank",
    "winner_rank_points",
    "loser_rank",
    "loser_rank_points",
]
```

```python
dfs = [pd.read_csv(f, usecols = KEEP_COLS) for f in files_1990]
df = pd.concat(dfs, ignore_index = True)
df.shape
```

```python
df.isna().sum()
```

```python
df = df.sort_values(["tourney_date", "tourney_id", "match_num"]).reset_index(drop = True)
df.head()
```

```python
df.to_csv(INTERIM_DIR / "atp_matches_1990_onward_base.csv", index = False)
```
