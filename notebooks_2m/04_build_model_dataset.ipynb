```python
import pandas as pd
from pathlib import Path

INTERIM_DIR = Path("../data/interim")
PROCESSED_DIR = Path("../data/processed")
```

```python
df = pd.read_csv(INTERIM_DIR / "atp_all_matches_with_features.csv")
df.shape
```

```python
df.head()
```

```python
df = df[df["tourney_date"] >= "1990-01-01"]
df.head()
```

```python
KEEP_CAT = ["G", "M", "A", "F"]
df = df[df["tourney_level"].isin(KEEP_CAT)]
print(df["tourney_level"].unique())
```

```python
df = df.sort_values(["tourney_date", "tourney_id", "match_num"])
df.columns.to_list()
```

```python
player_win = pd.DataFrame({
    "tourney_date": df["tourney_date"],
    "tourney_id": df["tourney_id"],
    "surface": df["surface"],
    "tourney_level": df["tourney_level"],
    "best_of": df["best_of"],
    "match_num": df["match_num"],

    "player_id": df["winner_id"],
    "player_age": df["winner_age"],
    "player_rank": df["winner_rank"],
    "player_rank_points": df["winner_rank_points"],
    "player_elo": df["winner_elo"],
    "player_surface_elo": df["winner_surface_elo"],
    "player_recent_form": df["winner_recent_form"],
    "player_recent_matches": df["winner_recent_matches"],
    "player_surface_relevance": df["winner_surface_relevance"],
    


    "opponent_id": df["loser_id"],
    "opponent_age": df["loser_age"],
    "opponent_rank": df["loser_rank"],
    "opponent_rank_points": df["loser_rank_points"],
    "opponent_elo": df["loser_elo"],
    "opponent_surface_elo": df["loser_surface_elo"],
    "opponent_recent_form": df["loser_recent_form"],
    "opponent_recent_matches": df["loser_recent_matches"],
    "opponent_surface_relevance": df["loser_surface_relevance"],

    "target": 1
})
```

```python
player_loss = pd.DataFrame({
    "tourney_date": df["tourney_date"],
    "tourney_id": df["tourney_id"],
    "surface": df["surface"],
    "tourney_level": df["tourney_level"],
    "best_of": df["best_of"],
    "match_num": df["match_num"],

    "player_id": df["loser_id"],
    "player_age": df["loser_age"],
    "player_rank": df["loser_rank"],
    "player_rank_points": df["loser_rank_points"],
    "player_elo": df["loser_elo"],
    "player_surface_elo": df["loser_surface_elo"],
    "player_recent_form": df["loser_recent_form"],
    "player_recent_matches": df["loser_recent_matches"],
    "player_surface_relevance": df["loser_surface_relevance"],
    


    "opponent_id": df["winner_id"],
    "opponent_age": df["winner_age"],
    "opponent_rank": df["winner_rank"],
    "opponent_rank_points": df["winner_rank_points"],
    "opponent_elo": df["winner_elo"],
    "opponent_surface_elo": df["winner_surface_elo"],
    "opponent_recent_form": df["winner_recent_form"],
    "opponent_recent_matches": df["winner_recent_matches"],
    "opponent_surface_relevance": df["winner_surface_relevance"],

    "target": 0
})
```

```python
model_df = pd.concat([player_win, player_loss], ignore_index=True)
model_df = model_df.sort_values(["tourney_id", "tourney_date", "match_num"])
model_df.shape
```

```python
model_df.head()
```

```python
import numpy as np

model_df["elo_diff"] = model_df["player_elo"] - model_df["opponent_elo"]
model_df["surface_elo_diff"] = model_df["player_surface_elo"] - model_df["opponent_surface_elo"]
model_df["age_diff"] = model_df["player_age"] - model_df["opponent_age"]
model_df["rank_diff"] = model_df["opponent_rank"] - model_df["player_rank"]
model_df["rank_log_diff"] = np.log(model_df["opponent_rank"]) - np.log(model_df["player_rank"])
model_df["rank_log_points_diff"] = np.log(model_df["player_rank_points"]) - np.log(model_df["opponent_rank_points"])
model_df["recent_form_diff"] = model_df["player_recent_form"] - model_df["opponent_recent_form"]
model_df["recent_matches_diff"] = model_df["player_recent_matches"] - model_df["opponent_recent_matches"]
model_df["surface_relevance_diff"] = model_df["player_surface_relevance"] - model_df["opponent_surface_relevance"]
```

```python
train_df = model_df[model_df["tourney_date"] < "2020-01-01"]
test_df = model_df[model_df["tourney_date"] >= "2020-01-01"]
```

```python
FEATURE_COLS = [
    # contesto partita
    "surface",
    "tourney_level",
    "best_of",
    
    # età
    "player_age",
    "opponent_age",
    "age_diff",
    
    # ranking assoluto + differenza log
    "player_rank",
    "opponent_rank",
    "rank_diff",
    "rank_log_diff",
    "rank_log_points_diff",
    
    # elo generale assoluto + differenza
    "player_elo",
    "opponent_elo",
    "elo_diff",
    
    # elo surface assoluto + differenza
    "player_surface_elo",
    "opponent_surface_elo",
    "surface_elo_diff",
    
    # recent form
    "player_recent_form",
    "opponent_recent_form",
    "recent_form_diff",
    
    # volume forma recente
    "player_recent_matches",
    "opponent_recent_matches",
    "recent_matches_diff",
    
    # rilevanza superficie nella forma recente
    "player_surface_relevance",
    "opponent_surface_relevance",
    "surface_relevance_diff",
]
```

```python
model_df[FEATURE_COLS + ["target"]].isna().sum()
```

```python
# rank mancante -> giocatore sconosciuto, assumiamo fondo classifica
train_df["player_rank"] = model_df["player_rank"].fillna(2500)
train_df["opponent_rank"] = model_df["opponent_rank"].fillna(2500)

# ricalcola rank_log_diff dopo imputazione
# np.log(2500) per i NaN che erano causati da rank mancante
train_df["rank_log_diff"] = np.log(model_df["player_rank"]) - np.log(model_df["opponent_rank"])

# age mancante -> imputiamo con la media
# l'età media nel tennis professionistico è intorno ai 25-26 anni
mean_age = model_df["player_age"].mean()
train_df["player_age"] = model_df["player_age"].fillna(mean_age)
train_df["opponent_age"] = model_df["opponent_age"].fillna(mean_age)
train_df["age_diff"] = model_df["player_age"] - model_df["opponent_age"]

# recent form e surface relevance -> lasciamo NaN
# XGBoost/LightGBM li gestiscono nativamente

# rank mancante -> giocatore sconosciuto, assumiamo fondo classifica
test_df["player_rank"] = model_df["player_rank"].fillna(2500)
test_df["opponent_rank"] = model_df["opponent_rank"].fillna(2500)

# ricalcola rank_log_diff dopo imputazione
# np.log(2500) per i NaN che erano causati da rank mancante
test_df["rank_log_diff"] = np.log(model_df["player_rank"]) - np.log(model_df["opponent_rank"])

# age mancante -> imputiamo con la media
# l'età media nel tennis professionistico è intorno ai 25-26 anni
mean_age = model_df["player_age"].mean()
test_df["player_age"] = model_df["player_age"].fillna(mean_age)
test_df["opponent_age"] = model_df["opponent_age"].fillna(mean_age)
test_df["age_diff"] = model_df["player_age"] - model_df["opponent_age"]

# recent form e surface relevance -> lasciamo NaN
# XGBoost/LightGBM li gestiscono nativamente
```

```python
train_df[FEATURE_COLS + ["target"]].to_csv(PROCESSED_DIR / "train.csv", index=False)
test_df[FEATURE_COLS + ["target"]].to_csv(PROCESSED_DIR / "test.csv", index=False)

print(f"Train: {train_df.shape}, Test: {test_df.shape}")
```
