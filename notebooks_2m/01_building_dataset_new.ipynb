```python
import pandas as pd
from pathlib import Path
```

```python
## definiamo i path verso le directory e creiamo se non esistono

DATA_DIR = Path("../data/raw/sackmann")
INTERIM_DIR = Path("../data/interim")
INTERIM_DIR.mkdir(parents = True, exist_ok = True)
```

```python
## racogliamo i file atp challenger e futures

all_files = sorted(DATA_DIR.glob("atp_matches_*.csv"))
```

```python
## carichiamo in dataframe e concateniamo i nostri vari csv
##usiamo low_mem per evitare problemi con rows vuote

dfs = [pd.read_csv(f, low_memory=False) for f in all_files]

df = pd.concat(dfs, ignore_index=True)

df = df.sort_values(["tourney_date", "tourney_id", "match_num"]).reset_index(drop=True)

print(df.shape)
df.head()
```

```python
df.to_csv(INTERIM_DIR / "atp_all_matches.csv", index = False)
```
